# Workbench CU — Colab acceptance fixture
Run cells individually. Cell 1 creates a unique run ID. Configure completion as `CU_DONE <run_id>` and artifact `result-<run_id>.json`, never as source literals. Cell 2 requests a local file, then text. Re-run cell 1 before each attempt. This notebook has not yet been validated in a live Colab session.

```python
import uuid, json, time
from pathlib import Path
run_id = uuid.uuid4().hex
print('CU_RUN', run_id)
print('Expected artifact:', f'result-{run_id}.json')
```

```python
from google.colab import files
from IPython.display import clear_output
print('CU_WAIT_FILE', run_id, flush=True)
uploaded = files.upload()
clear_output(wait=True)
print('CU_WAIT_TEXT', run_id, flush=True)
description = input('Mô tả tiếng Việt (nhập FAIL để thử lỗi): ')
clear_output(wait=True)
if description.strip() == 'FAIL':
    print('CU_FAILED', run_id, flush=True)
    raise RuntimeError('Intentional fixture failure')
for step in range(3):
    print('CU_PROGRESS', run_id, step + 1, flush=True)
    time.sleep(1)
artifact = Path(f'result-{run_id}.json')
artifact.write_text(json.dumps({'run_id': run_id, 'description': description, 'files': list(uploaded)}, ensure_ascii=False), encoding='utf-8')
from IPython.display import display, FileLink
display(FileLink(str(artifact)))
print('CU_DONE', run_id, flush=True)
```
